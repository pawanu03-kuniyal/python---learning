```python
# question 8 , Write a for loop to find the factorial of a given number.

def factorial():
    num = int(input("Enter a number: "))

    result = 1

    for i in range(1, num + 1):
        result = result * i

    print("Factorial:", result)

factorial()
```

```python
# question 8 ,

def fact():
    num = int(input("Enter your number: "))
    factorial = 1
    for i in range(1, num + 1):
        factorial = factorial * i

        print("Factorial =", factorial)

fact()
```

```python
# question 10 , Write a for loop to find the largest number in a list.

def largest_number():
    numbers = [10, 45, 23, 89, 12]

    largest = numbers[0]

    for num in numbers:
        if num > largest:
            largest = num

    print("Largest:", largest)

largest_number()
```

```python
# question 11 , Write a for loop to count the number of elements in a list.

def count_elements():
    numbers = [10, 20, 30, 40, 50]

    count = 0

    for num in numbers:
        count = count + 1

    print("Count:", count)

count_elements()
```

```python
# question 13 ,Write a for loop to reverse a string.

def reverse_string():
    text = "Python"
    reverse = ""

    for char in text:
        reverse = char + reverse

    print(reverse)

reverse_string()
```

```python
# question 14 , Write a for loop to count vowels in a string.

def count_vowels():
    text = "python programming"
    count = 0

    for char in text:
        if char in "aeiou":
            count = count + 1

    print("Vowels:", count)

count_vowels()
```

```python
# question 15 , Write a for loop to print numbers from 1 to 20 but stop when the number is 12. Use break.

def stop_at_12():
    for i in range(1, 21):
        if i == 12:
            break

        print(i)

stop_at_12()
```

```python
# question 16 , Write a for loop to print numbers from 1 to 20 but skip multiples of 4. Use continue.

def skip_multiples():
    for i in range(1, 21):
        if i % 4 == 0:
            continue

        print(i)

skip_multiples()
```

```python
# question 17 , Write a for loop to search for a number in a list and stop when it is found. Use break.

def search_number():
    numbers = [10, 20, 30, 40, 50]
    search = int(input("Enter number to search: "))

    for num in numbers:
        if num == search:
            print("Number found!")
            break
    else:
        print("Number not found.")


search_number()
```

```python
# question 18 , Write a for loop to print characters of a string but skip vowels. Use continue.

def skip_vowels():
    text = "Python Programming"

    for char in text:
        if char.lower() in "aeiou":
            continue

        print(char)


skip_vowels()
```

```python
# question 19 , Write a for loop to print numbers from 1 to 50 and stop when a number divisible by 9 is found. Use break.

def divisible_by_9():
    for i in range(1, 51):
        if i % 9 == 0:
            break

        print(i)


divisible_by_9()
```
